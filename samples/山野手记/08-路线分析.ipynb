# 山野手记 · 周末路线分析

用真实 Python 内核整理三条路线，给这周的出发留一个清晰的选择。

```python
routes = [
    ('溪谷轻徒步', 3.2, 120),
    ('山脊日落线', 6.8, 460),
    ('环山摄影线', 10.5, 680),
]
for name, distance, ascent in routes:
    print(f'{name}  |  {distance:.1f} km  |  爬升 {ascent} m')
```

```python
total = sum(distance for _, distance, _ in routes)
print(f'三条路线总里程：{total:.1f} km')
print('本周选择：山脊日落线，预留 3 小时，带相机与手记本。')
```
